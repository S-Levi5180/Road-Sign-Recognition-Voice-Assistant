```python
import xml.etree.ElementTree as ET
import os

# Paths
annotations_dir = "dataset/Annotations/Annotations"
images_dir = "dataset/Batch-1/batch"

# Pick an example XML file
xml_file = os.path.join(annotations_dir, "Datacluster Traffic Sign (1).xml")

# Parse XML
tree = ET.parse(xml_file)
root = tree.getroot()

# Get image filename
filename = root.find("filename").text
print(f"Image File: {filename}")

# Extract bounding box details
for obj in root.findall("object"):
    label = obj.find("name").text
    bbox = obj.find("bndbox")

    xmin0 = (bbox.find("xmin").text)
    value = bbox.find("xmin").text
    # Base DType extracted
    print(f"Raw value: {value}, Type: {type(value)}")
    xmin0 = float(xmin0)
    # Float Conversion
    print(f"After float conversion: {xmin0}, Type: {type(xmin0)}")
    # Int conversion
    xmin0 = int(xmin0)
    print(f"After int conversion: {xmin0}, Type: {type(xmin0)}")
    
    ymin0 = int(float(bbox.find("ymin").text))
    xmax0 = int(float(bbox.find("xmax").text))
    ymax0 = int(float(bbox.find("ymax").text))

    print(f"Label: {label}, BBox: ({xmin0}, {ymin0}, {xmax0}, {ymax0})")
```

```python
import cv2
import matplotlib.pyplot as plt

# Load image
image_path = os.path.join(images_dir, filename)
image = cv2.imread(image_path)

# Convert BGR to RGB
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Draw bounding box
cv2.rectangle(image, (xmin0, ymin0), (xmax0, ymax0), (255, 0, 0), 2)

# Show image
plt.imshow(image)
plt.title(label)
plt.axis("off")
plt.show()
```

```python
for xml_file in os.listdir(annotations_dir):
    if xml_file.endswith(".xml"):
        tree = ET.parse(os.path.join(annotations_dir, xml_file))
        root = tree.getroot()
        filename = root.find("filename").text
        image_path = os.path.join(images_dir, filename)

        # Load image
        image = cv2.imread(image_path)
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

        # Get bounding boxes
        for obj in root.findall("object"):
            label = obj.find("name").text
            bbox = obj.find("bndbox")
            xmin, ymin, xmax, ymax = [int(float(bbox.find(coord).text)) for coord in ("xmin", "ymin", "xmax", "ymax")]

            # Draw bounding box
            cv2.rectangle(image, (xmin, ymin), (xmax, ymax), (255, 0, 0), 2)

        # Show image
        plt.imshow(image)
        plt.title(f"{filename} - {label}")
        plt.axis("off")
        plt.show()
```
